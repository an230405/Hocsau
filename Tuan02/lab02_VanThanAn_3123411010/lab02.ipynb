Bài 7. Các hàm thống kê trên Tensor

Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

Yêu cầu:

• Thay dim=0 bằng dim=1 và nhận xét sự khác nhau.

dim=0  xử lý theo cột, x.sum(dim=0): Cộng hàng 0 + hàng 1 + hàng 2... ứng với mỗi cột Thu được tổng của từng cột.

dim=1 xử lý theo hàng, x.sum(dim=1): Cộng cột 0 + cột 1 + cột 2... ứng với mỗi hàng Thu được tổng của từng hàng.

```python
import torch 
 
a = torch.randn(3, 4) 
 
print(a) 
print("\nMean(Trung bình):") 
print(torch.mean(a, dim=0)) 
 
print("\nStandard deviation(Độ lệch chuẩn):") 
print(torch.std(a, dim=0)) 
 
print("\nCumulative sum(Tổng tích lũy):") 
print(torch.cumsum(a, dim=0)) 
```

Bài 8. Autograd cơ bản

Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

Yêu cầu:

• Tính thủ công đạo hàm của y = x² tại x = 3.6 và so sánh với x.grad.

• Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

```python
import torch 
 
x = torch.tensor( 
    3.6, 
    requires_grad=True 
) 
 
y = x * x 
 
y.backward() 
 
print("x =", x) 
print("y =", y) 
print("x.grad =", x.grad) 
print("Đạo hàm thủ công:",3.6*2)
x.grad.zero_()

y = 3*x**2 + 2*x + 1
y.backward()
print("Đổi hàm thành: ",y)
print("x.grad mới:", x.grad)
```

Bài 9. Gradient Descent tìm hệ số đa thức

Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

Yêu cầu:

• So sánh ba hệ số học được với [1, 2, 3].

• Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().

```python
import numpy as np
import torch

# Tạo đa thức
polynomial = np.poly1d([1, 2, 3])

N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

# Tạo ma trận đặc trưng
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
]) 
 
w = torch.randn(3, 1, requires_grad=True) 
x = torch.tensor(XX, dtype=torch.float32) 
y = torch.tensor(Y, dtype=torch.float32) 
 
optimizer = torch.optim.NAdam([w], lr=0.01) 
 
print("Initial coefficients(Hệ số ban đầu):") 
print(w) 
 
for epoch in range(1000): 
    optimizer.zero_grad() # Xóa gradient cũ 
    y_pred = x @ w 
    mse = torch.mean(torch.square(y - y_pred)) 
    mse.backward() # Tính gradient mới, lưu vào w.grad
    optimizer.step() # Dùng gradient để cập nhật w
 
print("\nLearned coefficients(Hệ số đã học):") 
print(w) 
```

Bài 10. Giải hệ phương trình bằng Autograd

Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

Yêu cầu:

• Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.

• Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True) # tạo ra một số ngẫu nhiên và cho phép tính gradient sau này
B = torch.tensor(random.random(), requires_grad=True)   
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000 # tạo ra 2000 lần lặp để tính toán giá trị tối ưu

optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01) # tạo ra một đối tượng optimizer NAdam để sửa A, B, C, D với learning rate = 0.01
for _ in range(EPOCHS): # lập lại quá trình 2000 lần
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4 # tính tổng biến động của y1, y2, y3, y4 và lấy bình phương của nó để tính sqerr
    
    optimizer.zero_grad() # reset gradient về 0 để gradient cũ không bị cộng vào
    sqerr.backward() # tính gradient của sqerr đối với A, B, C, D
    optimizer.step() # cập nhật giá trị của A, B, C, D dựa trên gradient vừa tính được
    
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:

• Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.

• Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A\*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.

• Cho y = 2x² + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.

• Thay đa thức trong Bài 9 bằng y = 2x² - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import torch
import numpy as np

# 1. Tạo Tensor X có 24 phần tử và reshape
X = torch.arange(1, 25)

print("Tensor X ban đầu:", X)
print("\nX reshape 4 x 6:\n", X.reshape(4, 6))
print("\nX reshape 2 x 12:\n", X.reshape(2, 12))
print("\nX reshape 2 x 3 x 4:\n", X.reshape(2, 3, 4))

# 2. Tạo hai Tensor A, B kích thước 3 x 4 và tính toán thống kê
A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("\nTensor A:\n", A)
print("\nTensor B:\n", B)
print("\nA + B:\n", A + B)
print("\nA * B:\n", A * B)
print("\nA / B:\n", A / B)

# Trung bình theo cột (dim=0)
print("\nTrung bình theo cột A:\n", torch.mean(A, dim=0))
print("\nTrung bình theo cột B:\n", torch.mean(B, dim=0))

# Độ lệch chuẩn theo cột (dim=0)
print("\nĐộ lệch chuẩn theo cột A:\n", torch.std(A, dim=0))
print("\nĐộ lệch chuẩn theo cột B:\n", torch.std(B, dim=0))

# 3. Autograd: y = 2x^2 + 5x + 3 tại x = 2
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()

print("\nx =", x.item())
print("y =", y.item())
print("Đạo hàm bằng Autograd:", x.grad.item())
print("Đạo hàm tính thủ công (4*x + 5):", 4 * x.item() + 5)

# 4. Huấn luyện đa thức: y = 2x^2 - 3x + 5
polynomial = np.poly1d([2, -3, 5])

# Tạo dữ liệu
N = 20
X_np = np.random.randn(N, 1) * 5
Y_np = polynomial(X_np)

# Tạo ma trận đặc trưng [X^2, X, 1]
XX = np.hstack([
    X_np ** 2,
    X_np,
    np.ones_like(X_np)
])

# Chuyển sang Tensor
x_train = torch.tensor(XX, dtype=torch.float32)
y_train = torch.tensor(Y_np, dtype=torch.float32)

# Khởi tạo trọng số ngẫu nhiên w
w = torch.randn(3, 1, requires_grad=True)

optimizer = torch.optim.NAdam([w], lr=0.03)

for epoch in range(1000):
    optimizer.zero_grad() # Xóa gradient cũ 
    y_pred = x_train @ w
    loss = torch.mean(torch.square(y_train - y_pred))
    loss.backward() # Tính gradient mới, lưu vào w.grad
    optimizer.step() # Dùng gradient để cập nhật w

print("\nKết quả huấn luyện đa thức:")
print("Hệ số x^2 =", w[0].item())
print("Hệ số x =", w[1].item())
print("Hệ số tự do =", w[2].item())
```
